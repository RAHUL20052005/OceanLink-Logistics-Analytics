```python
import pandas as pd

df = pd.read_excel("Delivery_Tracking.xlsx")
print("Rows before:", len(df))                      # 60141

# 1. Sort so "first" means the lowest Tracking_ID
df = df.sort_values("Tracking_ID")

# 2. Number the events within each shipment: 0, 1, 2, 3, 4, 5
df["event_no"] = df.groupby("Shipment_ID").cumcount()

# 3. Rows numbered 3 or higher are the extra copies
extra = df[df["event_no"] >= 3]
print("Duplicates found:", len(extra))              # 141

# 4. Save them before deleting
extra.drop(columns="event_no").to_excel("removed_duplicates_tracking.xlsx", index=False)

# 5. Keep only the first 3 events per shipment
df = df[df["event_no"] < 3].drop(columns="event_no")
print("Rows after:", len(df))                       # 60000
```

```python
dd
```

```python
# 1. Sort so "first" means the lowest Tracking_ID
dd = dd.sort_values("Tracking_ID")

# 2. Number the events within each shipment: 0, 1, 2, 3, 4, 5
dd["event_no"] = dd.groupby("Shipment_ID").cumcount()

# 3. Rows numbered 3 or higher are the extra copies
extra = dd[dd["event_no"] >= 3]
print("Duplicates found:", len(extra))              # 141

# 4. Save them before deleting
extra.drop(columns="event_no").to_excel("removed_duplicates_tracking.xlsx", index=False)

# 5. Keep only the first 3 events per shipment
dd = dd[dd["event_no"] < 3].drop(columns="event_no")
print("Rows after:", len(dd))                       # 60000
```

```python
# Convert the date column (invalid values become NaT)
dd["Delivery_Date"] = pd.to_datetime(dd["Delivery_Date"], errors="coerce")

# Trim and standardise the text columns
dd["Delivery_Status"] = dd["Delivery_Status"].str.strip().str.title()
dd["Current_Location"] = dd["Current_Location"].str.strip()

# Check the nulls (leave them as they are)
print(dd.isna().sum())          # Delivery_Date should show about 1,069 nulls

# Save the cleaned file
dd.to_csv("Delivery_Tracking_clean.csv", index=False)
```
